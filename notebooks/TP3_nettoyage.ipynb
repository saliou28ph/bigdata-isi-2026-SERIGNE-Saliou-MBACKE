### 3.1 Manquants et email


````python
from pyspark.sql import DataFrame

def unifier_manquants(df: DataFrame) -> DataFrame:
    """Emails "" / "N/A" -> null."""
    e = F.trim(F.col("email"))
    return df.withColumn(
        "email",
        F.when(e.isin("", "N/A", "n/a", "NULL"), None).otherwise(e))

def normaliser_email(df: DataFrame) -> DataFrame:
    """Email en minuscules + trim ; drapeau de validite."""
    motif = r"^[a-z0-9._%+-]+@[a-z0-9.-]+\.[a-z]{2,}$"
    df = df.withColumn("email", F.lower(F.trim(F.col("email"))))
    return df.withColumn(
        "email_valide",
        F.when(F.col("email").isNull(), F.lit(None))
         .otherwise(F.col("email").rlike(motif)))
````

### 3.2 Ville (avec retrait d'accents)


````python
import unicodedata
from pyspark.sql.types import StringType

def sans_accent(s):
    if s is None:
        return None
    nfkd = unicodedata.normalize("NFKD", s)
    return "".join(c for c in nfkd if not unicodedata.combining(c))

sans_accent_udf = F.udf(sans_accent, StringType())

def normaliser_ville(df: DataFrame) -> DataFrame:
    """ville (affichage) + ville_norm (cle sans accent)."""
    df = df.withColumn("ville", F.initcap(F.trim(F.col("ville"))))
    return df.withColumn(
        "ville_norm",
        F.lower(sans_accent_udf(F.trim(F.col("ville")))))
````

### 3.3 Téléphone et date de naissance


````python
def normaliser_telephone(df: DataFrame) -> DataFrame:
    """9 chiffres, prefixe 70/75/76/77/78 ; drapeau de validite."""
    tel = F.regexp_replace(F.col("telephone"), r"[^0-9]", "")
    tel = F.regexp_replace(tel, r"^221", "")
    return (df.withColumn("tel_norm", tel)
              .withColumn("tel_valide",
                  tel.rlike(r"^(70|75|76|77|78)\d{7}$")))

def valider_naissance(df: DataFrame) -> DataFrame:
    """Date plausible entre 1920 et aujourd'hui, sinon null."""
    d = F.to_date(F.col("date_naissance"), "yyyy-MM-dd")
    return df.withColumn(
        "date_naissance",
        F.when((d >= F.lit("1920-01-01")) & (d <= F.current_date()), d)
         .otherwise(None))
````

### 3.4 Déduplication (après normalisation)


````python
def dedupliquer_clients(df: DataFrame) -> DataFrame:
    """Doublons exacts puis 1 ligne par customer_id."""
    from pyspark.sql.window import Window
    w = Window.partitionBy("customer_id").orderBy(
        F.col("date_inscription").desc())
    return (df.dropDuplicates()
              .withColumn("_r", F.row_number().over(w))
              .filter(F.col("_r") == 1)
              .drop("_r"))
````

## 4. Assembler le pipeline et mesurer l'effet


````python
def nettoyer_clients(df: DataFrame) -> DataFrame:
    return (df
        .transform(unifier_manquants)
        .transform(normaliser_email)
        .transform(normaliser_ville)
        .transform(normaliser_telephone)
        .transform(valider_naissance)
        .transform(dedupliquer_clients))

df_net = nettoyer_clients(df_brut)

releves["villes_distinctes_apres"] = df_net.select("ville_norm").distinct().count()
releves["lignes_apres_nettoyage"]  = df_net.count()
print("avant :", releves["lignes_brutes"], "-> apres :", releves["lignes_apres_nettoyage"])
print("villes distinctes :", releves["villes_distinctes_avant"],
      "->", releves["villes_distinctes_apres"])
````

**Interprétation :**

La normalisation a donc principalement servi à rendre le format des noms plus homogène.

### 2. Traitement des valeurs manquantes

**Décision : `fill`**

Nous avons choisi de ne pas supprimer les enregistrements contenant une adresse e-mail absente. Les différentes formes utilisées pour représenter une absence de valeur, telles que `""`, `"N/A"`, `"n/a"` ou `"NULL"`, sont converties en `null`.

Cette approche permet de conserver les autres informations présentes dans les lignes concernées. Elle permet également d'obtenir une structure plus cohérente pour les données manquantes, ce qui facilite les traitements et les analyses ultérieurs avec PySpark.

### 3. Utilisation d'une UDF

Une UDF (*User Defined Function*) telle que `sans_accent` peut ralentir l'exécution d'un traitement Spark, notamment parce qu'elle nécessite l'exécution de code Python pour traiter les données. Les fonctions natives de Spark sont généralement plus performantes car elles sont mieux optimisées pour les traitements distribués.

Dans notre cas, l'utilisation de cette UDF reste pertinente puisqu'elle permet d'effectuer une opération personnalisée : retirer les accents présents dans les chaînes de caractères. Lorsque les fonctions natives ne permettent pas d'obtenir facilement le résultat recherché, une UDF constitue donc une solution adaptée.

### 4. Déduplication des données

Une déduplication effectuée directement sur les données recherche principalement les enregistrements présentant exactement les mêmes valeurs. Elle peut ainsi laisser passer certains doublons lorsque ceux-ci comportent de petites différences d'écriture.

Par exemple, une même ville peut apparaître sous plusieurs formes en raison d'une différence de casse, d'un espace supplémentaire ou de la présence d'un accent.

La méthode basée sur la normalisation consiste d'abord à mettre les valeurs dans un format commun, puis à effectuer la recherche des doublons. Cette approche permet ainsi d'identifier plus efficacement les enregistrements qui correspondent réellement à la même information malgré des différences de présentation.
